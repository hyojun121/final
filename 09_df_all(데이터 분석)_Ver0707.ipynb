## 📤 데이터를 불러온다.

```python
# 칼럼 84개인 전체 데이터 불러온다.
df_all = pd.read_csv("C:/Users/user/Desktop/workspace/14_Final_PROJECT/all_train_상관분석.csv")
```

```python
# ✅ 예측에 불필요하거나 타겟인 열 제거
X = df_all.drop(columns=['Segment', 'ID'])
y = LabelEncoder().fit_transform(df_all['Segment'])

# ✅ 범주형 인코딩
X_encoded = pd.get_dummies(X)

# ✅ 모델 학습
model = RandomForestClassifier()
model.fit(X_encoded, y)

# ✅ 중요도 계산
importances = pd.Series(model.feature_importances_, index=X_encoded.columns)

# ✅ 상위 50개 컬럼만 선택
top_50_cols = importances.nlargest(50).index.tolist()

# ✅ 해당 컬럼만 남긴 데이터프레임 구성
X_selected = X_encoded[top_50_cols]

# ✅ 타겟 y 다시 붙이기 (필요한 경우)
df_selected = pd.concat([X_selected, pd.Series(y, name='Segment')], axis=1)

# ✅ 확인
print("✅ 최종 남은 컬럼 수 (피처 + 타겟):", df_selected.shape[1])
print("📋 컬럼 목록:", df_selected.columns.tolist())
```
