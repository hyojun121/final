```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib as plt
plt.rcParams['font.family'] = 'AppleGothic'  # macOS에서 한글 폰트 적용
```

```python
df = pd.read_csv('csv/컬럼중요도30개 선택.csv')
```

```python
df = df.drop('ID.1', axis = 1)
```

```python
df.columns
```

```python
use_amount_cols = [
    '_2순위쇼핑업종_이용금액',
    '쇼핑_도소매_이용금액',
    '이용금액_할부_무이자_B0M',
    '이용금액_할부_무이자_R3M',
    '이용금액_일시불_B0M',
    '최대이용금액_할부_R12M',
    '최대이용금액_일시불_R12M',
    '_1순위업종_이용금액',
    '_1순위카드이용금액',
    '이용금액_할부_R12M',
    '이용금액_R3M_신용',
    '이용금액_R3M_신용체크',
    '이용금액_오프라인_R6M',
    '이용금액_CA_R12M',
    
]
```

### 이용금액 컬럼만

```python
df_A = df[df['Segment'] == 'A']

# 평균값 데이터프레임이 없다면 여기서 재생성
mean_values = df_A[use_amount_cols].mean()
mean_df = mean_values.reset_index()
mean_df.columns = ['컬럼명', '평균금액']

mean_values2 = df_A[use_amount_cols].median()
mean_df2 = mean_values2.reset_index()
mean_df2.columns = ['컬럼명', '중앙값']

# 시각화
fig = plt.figure(figsize=(12, 6))  # ✅ 여기서 plt.figure가 함수여야 함
sns.barplot(data=mean_df, x='컬럼명', y='평균금액', palette = 'mako')
sns.lineplot(data=mean_df2, x='컬럼명', y='중앙값', color='gray', marker='o', label='중앙값')
plt.title('Segment A - 이용금액 관련 컬럼별 분포', fontsize=16)
plt.ylabel('평균 금액')
plt.xlabel('이용금액 컬럼')
plt.xticks(rotation=45, ha='right')
plt.ylim(0, 200000)
plt.tight_layout()
plt.show()
```
